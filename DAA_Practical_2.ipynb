Practical 2a Linear Search

```python
import time

# Input from the user
n = int(input("Enter the number of elements: "))

arr = []
print(f"Enter {n} sorted elements:")
for i in range(n):
    arr.append(int(input()))

# Element to search
key = int(input("Enter the element to search: "))

# Start timer
start_time = time.time()

# Linear Search
position = -1
for i in range(n):
    if arr[i] == key:
        position = i
        break

# Stop timer
end_time = time.time()

# Calculate execution time
execution_time = end_time - start_time

# Display result
if position != -1:
    print(f"\nElement {key} found at index {position}")
else:
    print(f"\nElement {key} not found")

# Display execution time
print(f"\nExecution Time: {execution_time:.6f} seconds")

# Display time complexity
print("\nTime Complexity:")
print("Best Case    : O(1)")
print("Average Case : O(n)")
print("Worst Case   : O(n)")
print("Space Complexity: O(1)")
```

Practical 2b Binary Search

```python
import time

# Input from the user
n = int(input("Enter the number of elements: "))

arr = []
print(f"Enter {n} sorted elements:")
for i in range(n):
    arr.append(int(input()))

# Element to search
key = int(input("Enter the element to search: "))

# Start timer
start_time = time.time()

# Binary Search
low = 0
high = n - 1
position = -1

while low <= high:
    mid = (low + high) // 2

    if arr[mid] == key:
        position = mid
        break
    elif arr[mid] < key:
        low = mid + 1
    else:
        high = mid - 1

# Stop timer
end_time = time.time()

# Calculate execution time
execution_time = end_time - start_time

# Display result
if position != -1:
    print(f"\nElement {key} found at index {position}")
else:
    print(f"\nElement {key} not found")

# Display execution time
print(f"\nExecution Time: {execution_time:.6f} seconds")

# Display time complexity
print("\nTime Complexity:")
print("Best Case    : O(1)")
print("Average Case : O(log n)")
print("Worst Case   : O(log n)")
print("Space Complexity: O(1)")
```
